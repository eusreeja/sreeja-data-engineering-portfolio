## Start Spark

```python
from pyspark.sql import SparkSession

spark = (
    SparkSession.builder
    .appName("ClinVarIngestion")
    .master("local[*]")
    .getOrCreate()
)

spark
```

### Read ClinVar File

```python
raw_df = (
    spark.read
    .option("header", True)
    .option("sep", "\t")
    .csv("../data/raw/variant_summary.txt")
)
```

### Check Dataset Size

```python
print(f"Rows: {raw_df.count():,}")
print(f"Columns: {len(raw_df.columns)}")
```
